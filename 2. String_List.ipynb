### Unicode VS ASCII

# Creating a String

```python
# Creating Strings
s = 'hello'
print(s)
s1 = str('HELLO')
print(s1)
s2 = "World"
print(s2)

# for multiline string
s3 = '''Hi'''
print(s3)
s4 = """wow"""
print(s4)
```

# Accessing a String

```python
# Accessing Substrings from a Strings

# Indexing
## Positive
st = 'hello world'
print(st[2])
## Negative
print(st[-4])
```

# Slicing

```python
# Slicing - start inclusive & end exclusive
s0 = 'hello world'
print(s0[0:5])

print(s0[:3])

# step size
print(s0[0:6:2])
```

# Reverse a string

```python
print(s0[::-1])
```

```python
print(s0[-5:])   

print(s0[-1:-6:-1])
```

- ## Whenever you use negative indexing start would always be greater than end

## Adding Chars to Strings

- Conactenation (Using '+' )
- f-Strings (formatted String)
- Slicing

```python
# Adding Chars to Strings

# Concatenation
print('Concatenation')
text = "hello"
# Add to the end
new_text = text + "!"
print(new_text)
# Add to the start
new_text = "A " + text
print(new_text)

# f-Strings (formatted Strings)
print('Formatted Strings')
char = "X"
word = "marks the spot"
# Combine using f"..."
result = f"{char} {word}"
print(result)

# Slicing
print('Slicing')
word = "pyhon"  # Missing a 't'
char_to_add = "t"
# Slice before index 2, add 't', slice from index 2 onwards
fixed_word = word[:2] + char_to_add + word[2:]  # "python"
print(fixed_word)
```

# Operations on Strings

- Arithmetic
- Relational
- Logical
- Loops on Strings
- Membership 

### Arithmetic ( + * )

```python
# Arithmetic
print('delhi' + ' ' + 'mumbai')
print('delhi' * 5)
print('*' * 50)
```

### Relational ( == != > < >= <= )

```python
# Relational
'delhi' == 'mumbai'
'delhi' != 'mumbai'

# lexiographically compare strings (ASCII value comparision)
'mumbai' > 'pune'
'Pune' > 'pune'
```

### Logical  ( and or not )

```python
# Logical
print('AND')
print('hello' and 'world')
print('hello' or 'world')
print()
print('OR')
print('' and 'world')
print('' or 'world')
print()
print('NOT')
print(not 'hello')
```

- ## if string has characters then True else empty means False

### Loops on Strings

```python
# Loops on Strings
for i in 'hello':
    print(i, end = ' ')
```

```python
for i in 'delhi':
    print('pune')
```

### Membership  ( in  not in )

```python
# Membership
print('NOT IN')
print('D' not in 'delhi')
print()

print('IN')
print('D' in 'delhi')
```

### split

```python
print('Hi, my name is kashish'.split())
print('Hi, my name is kashish'.split('i'))
```

### join

```python
# reverse of split --> joins words to form sentence
print(' '.join(['Hi,', 'my', 'name', 'is', 'kashish']))
print('-'.join(['Hi,', 'my', 'name', 'is', 'kashish']))
```

### replace

```python
# replace given word with another (given word, another)
print('Hi, my name is kashish'.replace('Hi','hello'))
```
