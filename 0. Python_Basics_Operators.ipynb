# 6. Type Conversion 
## Process of changing a value from one data type (like an integer) to another (like a string or float). 

## Doesn't changes original data

```python
# Implicit type conversion (Automatic)
print(5+5.6)
print(type(5), type(5.6))

# Explicit type conversion (Forceful)
# print(4 + '4')  -- will not work
# str to int
print(4 + (int)('4'))

# int to str
str(5)

# int to float
float(4)
```

# 7. Literals
##  fixed value written directly in your source code. Unlike variables, which act as containers for data that can change, literals represent constant data that remains exactly as written during program execution. 

```python
# Types of Literals
# Binary
a = 0b1010

# Decimal
b = 100

# Octal
c = 0o310

# Hexadecimal
d = 0x12c

# Float
f1 = 10.5
f2 = 1.5e2
f3 = 1.5e-3

# Complex
x = 3.14j

print(a,b,c,d)
print(f1,f2,f3)
print(x, x.imag, x.real)
```

```python
# String
string = 'This is Python'
strings = "This is Python"
char = "C"
multiline_str = """This is a multiline string with 
                more than one line code"""
unicode = u"\U0001f600\U0001F606\U0001F923"
raw_str = r"raw \n string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
```

```python
# Boolean
a = True + 4
b = False + 10

print("a: ", a)
print("b: ", b)
```

# 8. Operators
## Special symbols or keywords used to perform computations on variables and values. The data they act upon is called an operand. For example, in the expression 10 + 5, the symbol + is the operator and the numbers 10 and 5 are the operands. 

# Types of Operators

```python
# - Arithmetic : + - * / (float) //(integer) %(modulo) **(power)
print(5+6)
print(5-6)
print(5*6)
print(5/6)
print(5//6)
print(5%6)
print(5**2)
```

```python
# - Relational : < <= > >= == === != !==
print(4>5)
print(4<5)
print(4>=5)
print(4<=5)
print(4==5)
print(4!=5)
```

```python
# - Logical : and(&&) or(||) not(!)
print(1 and 0)
print(1 or 0)
print(not 1)
```

```python
# - Bitwise : and(&) or(|) not(~) xor(^) left shift(<<) right shift (>>)
print(2 & 3)
print(2 | 3)
print(2 ^ 3)
print(~3)
print(4 << 2)
print(4 >> 2)
```

```python
# - Assignment : = += -= *= /= %=
a = 2
a += 2
print(a)
```

```python
# - Membership : in not in 
# case sensitive
print('D' in 'Delhi')
print(1 in [1,2,3,4,5,6])
print(2 not in (1,2,4,8,16))
```

```python
#  Program - Find the sum of a 3 digit number entered by the user
num = int(input("Enter a 3 digit number:"))

# 345 % 10 = 5
a = num % 10
num //= 10

# 34 % 10 = 4
b = num % 10
num //= 10

# 3 % 10 = 3
c = num % 10

print(a+b+c)
```
