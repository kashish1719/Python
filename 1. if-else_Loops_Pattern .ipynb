# 1. Branching
## Process of making decisions in your code. It allows a program to execute different blocks of statements based on whether a specific condition evaluates to True or False, giving the code the ability to react dynamically. 
### Rules 
### - Indentation is Mandatory: Python uses whitespace (indentation). Statements that belong to an if block must be indented (usually by 4 spaces).
### - No Parentheses Required: You don't need to wrap your conditions in parentheses.
### - Colon (:): Every branch statement header must end with a colon. 

# I. if
## Executes a block of code only if the condition evaluates to True. 

```python
age = 18
if age >= 18:
    print("You are eligible to vote.")
```

# II. if-else
## Provides two distinct paths: one block of code runs if the condition is True, and a fallback block runs if it is False.

```python
score = 45
if score >= 50:
    print("Pass")
else:
    print("Fail")
```

```python
# login program and indentation
# email -> demo@gmail.com
# password -> 1234
email = input('Enter email:')
password = input('Enter password')

if email == 'demo@gmail.com' and password == '1234':
    print('Login Successfull')
else:
    print('Not Found')
```

# III. if elif else
## Allows you to check multiple conditions in a sequence. It stops at the first condition that evaluates to True. 

```python
marks = 85
if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
else:
    print("Grade C")
```

# IV. nested if
## An if statement placed inside the body of another if or else block. This structure allows you to check for secondary conditions only after a primary condition has been satisfied, creating a multi-layered decision-making process.

```python
x = 41

if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

# V. match case
## A structural pattern matching tool that functions similarly to a switch statement, allowing you to branch based on multiple exact values. 

```python
status_code = 200
match status_code:
    case 200:
        print("Success")
    case 404:
        print("Not Found")
    case _:
        print("Unknown status")
```

 # While 
* Execute a block of code as long as a specific condition remains true. 
* Ideal when the number of iterations is not known beforehand.

```python
# program to print the table
num = int(input('Enter a number:'))

i = 1
while i <= 10:
    print(num , "*" , i , "=" , num*i)
    i += 1
```

```python
# sum of all digits of a given number
number = int(input('Enter a number:'))
sum = 0
while number > 0:
    sum += number%10
    number //= 10

print(sum)
```

```python
# keep accepting numbers from users till he/she enters a 0 and then find the avg
sum = count = 0   
n = int(input('Enter a number:'))

while n != 0:
    sum += n
    count += 1
    n = int(input('Enter a number:'))

if count > 0:
    print(sum/count)
else:
    print('No number entered')
```

```python
# while loop with else
x = 1

while x < 3:
    print(x)
    x += 1
else:
    print('Limit Exceeded')
```

# Guessing game

```python
# Guessing game
# generate a random integer between 1 and 100
import random
jackpot = random.randint(1,100)
guess = int(input('Guess the number:'))

count = 1
while guess != jackpot:
    if guess < jackpot:
        print('Wrong! guess higher number')
    else:
        print('Wrong! guess lower number')
        
    guess = int(input('Guess the number:'))
    count += 1
else:
    print('Correct guess')
    print('attempts:', count)
```

# For 

* Used to iterate over a sequence (like a list, string, or range).
* Used when you know in advance how many times you need to repeat a task.

```python
for i in range(1,11):
    print(i)

print('Odd No:')
for j in range(1,11,2):
    print(j)

print('reverse:')
for k in range(10,0,-1):
    print(k)
```

# Nested Loops

```python
# Program - Unique combination of 1,2,3,4
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

```python
# Program - Pattern 1
# *
# **
# ***

rows = int(input('Enter no of rows:'))
for i in range(1,rows+1):
    for j in range(1,i+1):
        print('*', end = '')
    print()
```

```python
# Program - Pattern 1
# 1
# 121
# 12321
# 1234321

rows = int(input('Enter no of rows:'))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print(j, end = '')
    for k in range(i-1,0,-1):
        print(k, end = '')
    print()
```

# Loop Control Statement
* Break
* Continue
* Pass

## break

- Immediately terminates the current loop entirely
- Program execution jumps to the very next line of code outside the loop.
- Used when you have found what you were searching for or an error occurs and you need to exit a for or while loop prematurely.

Ex - In Database, searching for a particular value / row

```python
# break
for i in range(1,11):
    if i == 5:
        break
    print(i, end = ' ')
```

```python
#  take range input from user and print all prime number between it
lower = int(input('Enter lower range:'))
upper = int(input('Enter upper range:'))

for i in range(lower, upper+1):
    for j in range(2,i):
        if i % j == 0:
            break
    else:
        print(i, end = ' ')
```

## continue

-  Skips the rest of the code for the current iteration only
- Pprogram immediately jumps back to the top of the loop and starts the next iteration.
- Used to ignore specific unwanted values or edge cases while allowing the loop to continue running through the rest of the data. 

Ex - In ecoomerce website, where product is out of stock you doesn't want to display. 

```python
# continue
for i in range(1,11):
    if i == 5:
        continue
    print(i, end = ' ')
```

## pass

- A null operation (does absolutely nothing).
- Program execution simply moves to the next line of code.
- Serves as a temporary placeholder. In many languages (especially Python), empty blocks of code cause a syntax error. pass is used when a function, class, or if statement is required by syntax but you haven't written the logic for it yet.


```python
# pass
for i in range(1,11):
    pass
```
